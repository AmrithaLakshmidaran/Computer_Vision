```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
in_img = cv2.imread("shawn.jpeg")
if in_img is None:
    print("Image not found. Check the file path or format.")
else:
    # Convert BGR to RGB
    in_img_rgb = cv2.cvtColor(in_img, cv2.COLOR_BGR2RGB)

    # Split the channels
    r, g, b = in_img[:, :, 0], in_img[:, :, 1], in_img[:, :, 2]

    # Gamma correction and grayscale conversion
    gamma = 1.04
    r_const, g_const, b_const = 0.2126, 0.7156, 0.0722
    gray_img = r_const * (r ** gamma) + g_const * (g ** gamma) + b_const * (b ** gamma)

    # Plot the images
    fig = plt.figure(figsize=(12, 6))

    # Display BGR image
    img1 = fig.add_subplot(131)
    img1.imshow(cv2.cvtColor(in_img, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for proper display
    img1.set_title('BGR Image')

    # Display RGB image
    img2 = fig.add_subplot(132)
    img2.imshow(in_img_rgb)
    img2.set_title('Original (RGB) Image')

    # Display grayscale image
    img3 = fig.add_subplot(133)
    img3.imshow(gray_img, cmap=plt.cm.get_cmap('gray'))
    img3.set_title('Grayscale Image')

    # Show the plot
    plt.tight_layout()
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
in_image = cv2.imread("Lena.jpeg")
if in_image is None:
    print("Image not found. Check the file path or format.")
    exit()

# Section 1: Contrast and Brightness Adjustment
alpha = 2.7  # Contrast control
beta = 3     # Brightness control
adjusted_img = cv2.convertScaleAbs(in_image, alpha=alpha, beta=beta)

plt.figure(figsize=(10, 7))
plt.subplot(121)
plt.imshow(cv2.cvtColor(in_image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(122)
plt.imshow(cv2.cvtColor(adjusted_img, cv2.COLOR_BGR2RGB))
plt.title('Contrast Adjusted Image')
plt.axis('off')
plt.show()

# Section 2: Resizing the Image
new_width, new_height = 250, 250
resize_img = cv2.resize(in_image, (new_width, new_height))

plt.figure(figsize=(10, 7))
plt.subplot(121)
plt.imshow(cv2.cvtColor(in_image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(122)
plt.imshow(cv2.cvtColor(resize_img, cv2.COLOR_BGR2RGB))
plt.title('Resized Image')
plt.axis('off')
plt.show()

# Section 3: RGB and Grayscale Conversion
in_img_rgb = cv2.cvtColor(in_image, cv2.COLOR_BGR2RGB)
gray_image = cv2.cvtColor(in_image, cv2.COLOR_BGR2GRAY)

fig = plt.figure(figsize=(15, 5))
imag1 = fig.add_subplot(131)
imag1.imshow(cv2.cvtColor(in_image, cv2.COLOR_BGR2RGB))
imag1.set_title('BGR Image')
imag1.axis('off')

imag2 = fig.add_subplot(132)
imag2.imshow(in_img_rgb)
imag2.set_title('Original (RGB) Image')
imag2.axis('off')

imag3 = fig.add_subplot(133)
imag3.imshow(gray_image, cmap='gray')
imag3.set_title('Grayscale Image')
imag3.axis('off')

plt.tight_layout()
plt.show()

# Section 4: Gamma Correction and Grayscale Conversion
gamma = 1.04
r_const, g_const, b_const = 0.2126, 0.7156, 0.0722
r, g, b = in_image[:, :, 0], in_image[:, :, 1], in_image[:, :, 2]
gray_img = r_const * (r ** gamma) + g_const * (g ** gamma) + b_const * (b ** gamma)

fig = plt.figure(figsize=(12, 6))

img1 = fig.add_subplot(131)
img1.imshow(cv2.cvtColor(in_image, cv2.COLOR_BGR2RGB))
img1.set_title('BGR Image')
img1.axis('off')

img2 = fig.add_subplot(132)
img2.imshow(in_img_rgb)
img2.set_title('Original (RGB) Image')
img2.axis('off')

img3 = fig.add_subplot(133)
img3.imshow(gray_img, cmap='gray')
img3.set_title('Grayscale Image with Gamma Correction')
img3.axis('off')

plt.tight_layout()
plt.show()

# Section 5: Grayscale Histogram
# Calculate and plot the histogram
histogram = cv2.calcHist([gray_image], [0], None, [256], [0, 256])

plt.figure(figsize=(8, 5))
plt.plot(histogram, color='gray')  # Set the color to gray
plt.xlim([0, 256])  # Set x-axis limits
plt.title('Histogram of Grayscale Image')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
