```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image2 = cv2.imread("scene.png")
rgb_image = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)

transformation_matrix = np.array([
    [0.299, 0.587, 0.114],
    [0.596, -0.274, -0.321],
    [0.211, -0.523, 0.311]
])

reshaped_rgb = rgb_image.reshape((-1, 3))
yiq_image = np.dot(reshaped_rgb, transformation_matrix)
yiq_image = yiq_image.reshape(rgb_image.shape)

y_channel, i_channel, q_channel = cv2.split(yiq_image)

plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.title('YIQ Image')
plt.imshow(yiq_image)
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(y_channel, cmap='gray')
plt.title('Y Channel')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(i_channel, cmap='coolwarm')
plt.title('I Channel')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(q_channel, cmap='coolwarm')
plt.title('Q Channel')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

image_path ="holi.jpeg"
image = Image.open(image_path)

rgb_arr = np.array(image)
cmy_arr = 255 - rgb_arr
cmy_image = Image.fromarray(cmy_arr.astype('uint8'), 'RGB')

r_channel, g_channel, b_channel = image.split()

r_arr = np.array(r_channel)
g_arr = np.array(g_channel)
b_arr = np.array(b_channel)
c_channel = 255 - r_arr
m_channel = 255 - g_arr
y_channel = 255 - b_arr

plt.figure(figsize=(12, 4))

plt.subplot(1, 4, 1)
plt.imshow(cmy_image)
plt.title('CMY Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(c_channel, cmap='gray')
plt.title('Cyan Channel')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(m_channel, cmap='gray')
plt.title('Magenta Channel')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(y_channel, cmap='gray')
plt.title('Yellow Channel')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Load and convert image to RGB
img = cv2.imread('boat.jpeg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert RGB to HSV
hsvimg = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)

# Modify HSV channels
hue = 10
saturation = 50
value = 30

h, s, v = cv2.split(hsvimg)
h = cv2.add(h, hue)
s = cv2.add(s, saturation)
v = cv2.add(v, value)

# Merge modified HSV channels and convert back to RGB
img_hsv_modified = cv2.merge((h, s, v))
img_rgb_modified = cv2.cvtColor(img_hsv_modified, cv2.COLOR_HSV2RGB)

# Plot images
plt.figure(figsize=(15, 15))

plt.subplot(3, 3, 1)
plt.imshow(img)
plt.title('Original RGB Image')
plt.axis('off')

plt.subplot(3, 3, 2)
plt.imshow(h, cmap='gray')
plt.title('Hue Channel')
plt.axis('off')

plt.subplot(3, 3, 3)
plt.imshow(s, cmap='gray')
plt.title('Saturation Channel')
plt.axis('off')

plt.subplot(3, 3, 4)
plt.imshow(v, cmap='gray')
plt.title('Value Channel')
plt.axis('off')

plt.subplot(3, 3, 5)
plt.imshow(cv2.cvtColor(img_hsv_modified, cv2.COLOR_HSV2RGB))
plt.title('Modified HSV Image')
plt.axis('off')

plt.subplot(3, 3, 6)
plt.imshow(img_rgb_modified)
plt.title('Modified RGB Image')
plt.axis('off')

plt.subplot(3, 3, 7)
plt.axis('off')

plt.subplot(3, 3, 8)
plt.axis('off')

plt.subplot(3, 3, 9)
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Load the image
image_path = "holi.jpeg"
image = Image.open(image_path).convert("RGB")
rgb_array = np.array(image)

# Manual RGB to CNY Conversion (Inversion of RGB channels)
cny_array_manual = np.zeros_like(rgb_array)
cny_array_manual[..., 0] = 255 - rgb_array[..., 0]  # Cyan = 255 - Red
cny_array_manual[..., 1] = 255 - rgb_array[..., 1]  # Magenta = 255 - Green
cny_array_manual[..., 2] = 255 - rgb_array[..., 2]  # Yellow = 255 - Blue

cny_image_manual = Image.fromarray(cny_array_manual)

# Built-in RGB to CNY Conversion (Using numpy to manually invert RGB)
cny_array_built_in = np.zeros_like(rgb_array)
cny_array_built_in[..., 0] = 255 - rgb_array[..., 0]  # Cyan = 255 - Red
cny_array_built_in[..., 1] = 255 - rgb_array[..., 1]  # Magenta = 255 - Green
cny_array_built_in[..., 2] = 255 - rgb_array[..., 2]  # Yellow = 255 - Blue

cny_image_built_in = Image.fromarray(cny_array_built_in)

# Display the original and both CNY images side by side
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Original image
axes[0].imshow(image)
axes[0].set_title("Original Image")
axes[0].axis('off')

# Built-in CNY image (manual matrix operation)
axes[1].imshow(cny_image_built_in)
axes[1].set_title("Built-in CMY")
axes[1].axis('off')

# Manual CNY image
axes[2].imshow(cny_image_manual)
axes[2].set_title("Manual CMY")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
